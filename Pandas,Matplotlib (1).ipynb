import pandas as pd

```python
import matplotlib.pyplot as plt
df.groupby('LocationID')['Attendees'].sum().plot(kind='bar')
plt.title('Total Attendees by Location')
plt.xlabel('Location ID')
plt.ylabel('Total Attendees')
```

```python
plt.plot(df.groupby('LocationID')['Attendees'].sum())
plt.title('Total Attendees by Location')
plt.xlabel('Location ID')
plt.ylabel('Total Attendees')
```

```python
plt.scatter(df["LocationID"],df["Attendees"],color="red",label="Points")
plt.title("Location ID vs Attendees")
plt.xlabel("Location ID")
plt.ylabel("Attendees")
plt.legend()

```

```python
location_attendees=df.groupby("LocationID")["Attendees"].sum()
plt.plot(location_attendees.index,location_attendees.values,marker="o",label="Attendees")

plt.plot(
    location_attendees.index,
    location_attendees.values,
    marker="o",
    label="Attendees"
)
plt.legend()
plt.grid()
plt.tight_layout()
```

```python
import pandas as pd
df=pd.read_excel(r"C:\Users\acer\Downloads\\power bi project\Gym Class Sheets.xlsx")
df.head()
```

```python
print(type(df))
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt
location_attendees=df.groupby("LocationID").agg({"Attendees":"sum"})
print(location_attendees)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(3,3))
plt.pie(
    location_attendees["Attendees"],
    labels=location_attendees.index,
    autopct="%.1f%%"
)
plt.title("Attendees Distribution by Location")
```
